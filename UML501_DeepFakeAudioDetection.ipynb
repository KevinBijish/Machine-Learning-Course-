```python
import os
import shutil
import sys
import numpy as np
import pandas as pd
import librosa
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import warnings

warnings.filterwarnings('ignore')

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

class Config:
    BASE_PATH = r"C:\Users\kevin\Downloads\archive (1)\LA\LA"
    PROTOCOL_PATH = os.path.join(BASE_PATH, "ASVspoof2019_LA_cm_protocols")
    TRAIN_AUDIO_PATH = os.path.join(BASE_PATH, "ASVspoof2019_LA_train", "flac")
    DEV_AUDIO_PATH = os.path.join(BASE_PATH, "ASVspoof2019_LA_dev", "flac")
    TRAIN_PROTOCOL = os.path.join(PROTOCOL_PATH, "ASVspoof2019.LA.cm.train.trn.txt")
    DEV_PROTOCOL = os.path.join(PROTOCOL_PATH, "ASVspoof2019.LA.cm.dev.trl.txt")
    SAMPLE_RATE = 16000
    DURATION = 4
    N_MFCC = 60
    N_FFT = 2048
    HOP_LENGTH = 512
    BATCH_SIZE = 64
    EPOCHS = 22
    LEARNING_RATE = 0.0001
    WEIGHT_DECAY = 1e-4
    NUM_WORKERS = 0
    HIDDEN_DIM = 512
    NUM_LAYERS = 4
    DROPOUT = 0.5
    USE_AUGMENTATION = True
    BALANCE_DATA = True

config = Config()

def load_cm_protocol(protocol_path, audio_dir):
    data = []
    n_missing = 0
    with open(protocol_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            _, file_name, _, _, key = parts
            label = 1 if key == 'bonafide' else 0
            flac_file = os.path.join(audio_dir, file_name + ".flac")
            if os.path.exists(flac_file):
                data.append({'file_name': file_name, 'label': label})
            else:
                n_missing += 1
    df = pd.DataFrame(data)
    bonafide_count = df['label'].sum() if len(df) > 0 else 0
    spoof_count = len(df) - bonafide_count
    print(f"Loaded {len(df)} samples. Missing audio files for {n_missing} entries.")
    print(f"Bonafide: {bonafide_count}, Spoof: {spoof_count}")
    return df

def extract_features(audio_path, sr=config.SAMPLE_RATE, duration=config.DURATION):
    try:
        y, sr = librosa.load(audio_path, sr=sr, duration=duration)
        target_length = sr * duration
        if len(y) < target_length:
            y = np.pad(y, (0, target_length - len(y)), mode='constant')
        else:
            y = y[:target_length]
        y = librosa.util.normalize(y)
        mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=config.N_MFCC,
                                   n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
        mfcc_delta = librosa.feature.delta(mfcc)
        mfcc_delta2 = librosa.feature.delta(mfcc, order=2)
        spectral_centroid = librosa.feature.spectral_centroid(y=y, sr=sr, n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
        spectral_rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr, n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
        zero_crossing = librosa.feature.zero_crossing_rate(y, hop_length=config.HOP_LENGTH)
        features = np.vstack([
            mfcc,
            mfcc_delta,
            mfcc_delta2,
            spectral_centroid,
            spectral_rolloff,
            zero_crossing
        ])
        return features
    except Exception as e:
        print(f"Error processing {audio_path}: {str(e)}")
        return np.zeros((config.N_MFCC * 3 + 3, 125))

def augment_features(features):
    aug_type = np.random.choice(['noise', 'time_mask', 'none'], p=[0.3, 0.3, 0.4])
    if aug_type == 'noise':
        noise = np.random.randn(*features.shape) * 0.005
        return features + noise
    elif aug_type == 'time_mask':
        features_copy = features.copy()
        t_max = features.shape[1]
        t_mask = int(t_max * 0.1)
        if t_max - t_mask <= 0:
            return features_copy
        t_start = np.random.randint(0, t_max - t_mask)
        features_copy[:, t_start:t_start+t_mask] = 0
        return features_copy
    return features

class ASVspoofDataset(Dataset):
    def __init__(self, protocol_df, audio_path, augment=False):
        self.protocol_df = protocol_df.reset_index(drop=True)
        self.audio_path = audio_path
        self.augment = augment
    def __len__(self):
        return len(self.protocol_df)
    def __getitem__(self, idx):
        row = self.protocol_df.iloc[idx]
        file_name = row['file_name'] + ".flac"
        label = row['label']
        audio_file = os.path.join(self.audio_path, file_name)
        features = extract_features(audio_file)
        if self.augment and np.random.random() > 0.5:
            features = augment_features(features)
        features = torch.FloatTensor(features)
        label = torch.LongTensor([label])
        return features, label

class ImprovedDeepfakeDetector(nn.Module):
    def __init__(self, input_dim=config.N_MFCC * 3 + 3, hidden_dim=config.HIDDEN_DIM,
                 num_layers=config.NUM_LAYERS, dropout=config.DROPOUT):
        super(ImprovedDeepfakeDetector, self).__init__()
        self.conv_layers = nn.Sequential(
            nn.Conv1d(input_dim, 256, kernel_size=3, padding=1),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.MaxPool1d(2),
            nn.Dropout(dropout * 0.5),
            nn.Conv1d(256, 512, kernel_size=3, padding=1),
            nn.BatchNorm1d(512),
            nn.ReLU(),
            nn.MaxPool1d(2),
            nn.Dropout(dropout * 0.5),
        )
        self.lstm = nn.LSTM(
            input_size=512,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0,
            bidirectional=True
        )
        self.attention = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, 1)
        )
        self.classifier = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(256, 2)
        )
    def forward(self, x):
        x = self.conv_layers(x)
        x = x.permute(0, 2, 1)
        lstm_out, _ = self.lstm(x)
        attention_weights = torch.softmax(self.attention(lstm_out), dim=1)
        context = torch.sum(attention_weights * lstm_out, dim=1)
        output = self.classifier(context)
        return output

def train_epoch(model, dataloader, criterion, optimizer, device):
    model.train()
    running_loss, all_preds, all_labels = 0.0, [], []
    for features, labels in tqdm(dataloader, desc="Training"):
        features, labels = features.to(device), labels.squeeze().to(device)
        optimizer.zero_grad()
        outputs = model(features)
        loss = criterion(outputs, labels)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        running_loss += loss.item()
        preds = torch.argmax(outputs, dim=1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
    epoch_loss = running_loss / len(dataloader) if len(dataloader) > 0 else 0.0
    epoch_acc = accuracy_score(all_labels, all_preds) if len(all_labels) > 0 else 0.0
    return epoch_loss, epoch_acc

def evaluate(model, dataloader, criterion, device):
    model.eval()
    running_loss, all_preds, all_labels, all_probs = 0.0, [], [], []
    with torch.no_grad():
        for features, labels in tqdm(dataloader, desc="Evaluating"):
            features, labels = features.to(device), labels.squeeze().to(device)
            outputs = model(features)
            loss = criterion(outputs, labels)
            running_loss += loss.item()
            probs = torch.softmax(outputs, dim=1)
            preds = torch.argmax(outputs, dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_probs.extend(probs[:, 1].cpu().numpy())
    epoch_loss = running_loss / len(dataloader) if len(dataloader) > 0 else 0.0
    epoch_acc = accuracy_score(all_labels, all_preds) if len(all_labels) > 0 else 0.0
    try:
        auc_score = roc_auc_score(all_labels, all_probs) if len(all_labels) > 0 else 0.0
    except:
        auc_score = 0.0
    return epoch_loss, epoch_acc, auc_score, all_preds, all_labels

def main():
    print("="*70)
    print("IMPROVED DEEPFAKE AUDIO DETECTION - ASVspoof 2019 LA")
    print("="*70)
    print("\n[1/6] Loading train protocol...")
    train_df = load_cm_protocol(config.TRAIN_PROTOCOL, config.TRAIN_AUDIO_PATH)
    if len(train_df) == 0:
        print("ERROR: No valid audio files found in training set.")
        return
    print("\n[2/6] Loading dev protocol...")
    dev_df = load_cm_protocol(config.DEV_PROTOCOL, config.DEV_AUDIO_PATH)
    if len(dev_df) == 0:
        print("ERROR: No valid audio files found in dev set.")
        return
    if config.BALANCE_DATA:
        print("\n[3/6] Balancing training data...")
        bonafide = train_df[train_df['label'] == 1]
        spoof = train_df[train_df['label'] == 0]
        print(f"Original - Bonafide: {len(bonafide)}, Spoof: {len(spoof)}")
        min_samples = min(len(bonafide), len(spoof))
        bonafide = bonafide.sample(n=min_samples, random_state=42)
        spoof = spoof.sample(n=min_samples, random_state=42)
        train_df = pd.concat([bonafide, spoof]).sample(frac=1, random_state=42).reset_index(drop=True)
        print(f"Balanced - Bonafide: {train_df['label'].sum()}, Spoof: {len(train_df) - train_df['label'].sum()}")
    print("\n[4/6] Creating datasets...")
    train_dataset = ASVspoofDataset(train_df, config.TRAIN_AUDIO_PATH, augment=config.USE_AUGMENTATION)
    dev_dataset = ASVspoofDataset(dev_df, config.DEV_AUDIO_PATH, augment=False)
    train_loader = DataLoader(train_dataset, batch_size=config.BATCH_SIZE, shuffle=True, num_workers=config.NUM_WORKERS)
    dev_loader = DataLoader(dev_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=config.NUM_WORKERS)
    print("\n[5/6] Initializing improved model...")
    model = ImprovedDeepfakeDetector().to(device)
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total parameters: {total_params:,}")
    print(f"Trainable parameters: {trainable_params:,}")
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.AdamW(model.parameters(), lr=config.LEARNING_RATE, weight_decay=config.WEIGHT_DECAY)
    scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2)
    history = {
        'train_loss': [], 'train_acc': [],
        'val_loss': [], 'val_acc': [], 'val_auc': []
    }
    best_val_acc = 0.0
    patience = 15
    patience_counter = 0
    print("\n[6/6] Starting training...")
    print("="*70)
    for epoch in range(config.EPOCHS):
        print(f"\n{'='*30} Epoch {epoch+1}/{config.EPOCHS} {'='*30}")
        train_loss, train_acc = train_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc, val_auc, val_preds, val_labels = evaluate(model, dev_loader, criterion, device)
        scheduler.step()
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        history['val_auc'].append(val_auc)
        print(f"\n📊 Metrics:")
        print(f"   Train Loss: {train_loss:.4f} | Train Acc: {train_acc*100:.2f}%")
        print(f"   Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc*100:.2f}%")
        print(f"   Val AUC:    {val_auc:.4f}")
        print(f"   Learning Rate: {optimizer.param_groups[0]['lr']:.6f}")
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_auc = val_auc

            checkpoint_state_path = 'best_model1_state.pth'
            temp_state_path = 'best_model1_state_temp.pth'

            try:
                torch.save(model.state_dict(), temp_state_path)
                loaded_state = torch.load(temp_state_path, map_location=device)
                verification_model = ImprovedDeepfakeDetector().to(device)
                verification_model.load_state_dict(loaded_state)
                shutil.move(temp_state_path, checkpoint_state_path)

                print(f"\n✅ Best model (state_dict) saved! Accuracy: {best_val_acc*100:.2f}%, AUC: {val_auc:.4f}")

            except Exception as e:
                print(f"\n❌ Error saving checkpoint (state_dict): {e}")
                if os.path.exists(temp_state_path):
                    os.remove(temp_state_path)

            
            try:
                meta = {
                    'epoch': epoch,
                    'val_acc': val_acc,
                    'val_auc': val_auc
                }
                import json
                with open('best_model1_meta.json', 'w') as jf:
                    json.dump(meta, jf)
            except Exception:
                pass

            patience_counter = 0
        else:
            patience_counter += 1
            print(f"   Patience: {patience_counter}/{patience}")
    print("\n📈 Generating training plots...")
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    axes[0].plot(history['train_loss'], label='Train Loss', linewidth=2)
    axes[0].plot(history['val_loss'], label='Val Loss', linewidth=2)
    axes[0].set_xlabel('Epoch', fontsize=12)
    axes[0].set_ylabel('Loss', fontsize=12)
    axes[0].legend(fontsize=11)
    axes[0].set_title('Training and Validation Loss', fontsize=13, fontweight='bold')
    axes[0].grid(True, alpha=0.3)
    axes[1].plot(history['train_acc'], label='Train Acc', linewidth=2)
    axes[1].plot(history['val_acc'], label='Val Acc', linewidth=2)
    axes[1].set_xlabel('Epoch', fontsize=12)
    axes[1].set_ylabel('Accuracy', fontsize=12)
    axes[1].legend(fontsize=11)
    axes[1].set_title('Training and Validation Accuracy', fontsize=13, fontweight='bold')
    axes[1].grid(True, alpha=0.3)
    axes[2].plot(history['val_auc'], label='Val AUC', linewidth=2)
    axes[2].set_xlabel('Epoch', fontsize=12)
    axes[2].set_ylabel('AUC Score', fontsize=12)
    axes[2].legend(fontsize=11)
    axes[2].set_title('Validation AUC Score', fontsize=13, fontweight='bold')
    axes[2].grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig('training_history.png', dpi=300, bbox_inches='tight')
    print("   Saved: training_history.png")
    print("\n" + "="*70)
    print("✅ TRAINING COMPLETED SUCCESSFULLY!")
    print("="*70)
    print(f"📁 Saved files:")
    print(f"   - best_model1_state.pth  (model.state_dict())")
    print(f"   - best_model1_meta.json  (training metadata)")
    print(f"   - training_history.png")
    print("="*70)

if __name__ == "__main__":
    main()
```

```python
import os, sys, traceback, gc
# ---------- Workarounds & thread limits (must be set BEFORE heavy imports) ----------
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
os.environ["OMP_NUM_THREADS"] = "1"
os.environ["OPENBLAS_NUM_THREADS"] = "1"
os.environ["MKL_NUM_THREADS"] = "1"

import numpy as np
import torch
torch.set_num_threads(1)
import matplotlib.pyplot as plt
from sklearn.metrics import (classification_report, confusion_matrix, 
                            accuracy_score, precision_score, recall_score, 
                            f1_score, roc_auc_score, roc_curve, 
                            precision_recall_curve, average_precision_score)
import seaborn as sns
try:
    import soundfile as sf
except Exception:
    sf = None
try:
    import resampy
except Exception:
    resampy = None
import librosa
from IPython.display import Audio, display

# ---------------- config (match training) ----------------
class Config:
    SAMPLE_RATE = 16000
    DURATION = 4
    N_MFCC = 60
    N_FFT = 2048
    HOP_LENGTH = 512

config = Config()

# ---------------- safe audio loading + feature extraction ----------------
def safe_load_audio(path, sr=config.SAMPLE_RATE, duration=config.DURATION):
    target_len = sr * duration
    y = None; loaded_sr = None

    if sf is not None:
        try:
            data, loaded_sr = sf.read(path, dtype='float32')
            if data is None:
                raise RuntimeError("soundfile returned None")
            if getattr(data, "ndim", 1) > 1:
                data = data.mean(axis=1)
            y = data.astype('float32')
        except Exception as e:
            print("soundfile failed:", e)
            y = None; loaded_sr = None

    if y is None:
        try:
            y, loaded_sr = librosa.load(path, sr=None, mono=True, duration=duration)
            y = y.astype('float32')
        except Exception as e:
            print("librosa load failed:")
            traceback.print_exc()
            raise RuntimeError("Unable to load audio. Install soundfile/libsndfile or ensure ffmpeg is available.") from e

    if loaded_sr != sr:
        if resampy is not None:
            y = resampy.resample(y, loaded_sr, sr)
            loaded_sr = sr
        else:
            try:
                y = librosa.resample(y, orig_sr=loaded_sr, target_sr=sr)
                loaded_sr = sr
            except Exception:
                raise RuntimeError("Need resampy or librosa resample support to convert sample rate.")

    if len(y) < target_len:
        y = np.pad(y, (0, target_len - len(y)), mode='constant')
    else:
        y = y[:target_len]

    maxv = np.max(np.abs(y)) if y.size else 1.0
    if maxv > 0:
        y = y / maxv
    return y, sr

def extract_features(audio_path, sr=config.SAMPLE_RATE, duration=config.DURATION):
    y, sr = safe_load_audio(audio_path, sr=sr, duration=duration)
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=config.N_MFCC,
                               n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
    mfcc_delta = librosa.feature.delta(mfcc)
    mfcc_delta2 = librosa.feature.delta(mfcc, order=2)
    spectral_centroid = librosa.feature.spectral_centroid(y=y, sr=sr, n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
    spectral_rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr, n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
    zero_crossing = librosa.feature.zero_crossing_rate(y, hop_length=config.HOP_LENGTH)
    features = np.vstack([mfcc, mfcc_delta, mfcc_delta2, spectral_centroid, spectral_rolloff, zero_crossing])
    return features

import torch.nn as nn
class ImprovedDeepfakeDetector(nn.Module):
    def __init__(self, input_dim=config.N_MFCC*3+3, hidden_dim=512, num_layers=4, dropout=0.5):
        super().__init__()
        self.conv_layers = nn.Sequential(
            nn.Conv1d(input_dim,256,3,padding=1), nn.BatchNorm1d(256), nn.ReLU(), nn.MaxPool1d(2), nn.Dropout(dropout*0.5),
            nn.Conv1d(256,512,3,padding=1), nn.BatchNorm1d(512), nn.ReLU(), nn.MaxPool1d(2), nn.Dropout(dropout*0.5),
        )
        self.lstm = nn.LSTM(input_size=512, hidden_size=hidden_dim, num_layers=num_layers,
                            batch_first=True, dropout=dropout if num_layers>1 else 0, bidirectional=True)
        self.attention = nn.Sequential(nn.Linear(hidden_dim*2, hidden_dim), nn.Tanh(), nn.Linear(hidden_dim,1))
        self.classifier = nn.Sequential(
            nn.Linear(hidden_dim*2, hidden_dim), nn.BatchNorm1d(hidden_dim), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(hidden_dim,256), nn.BatchNorm1d(256), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(256,2)
        )
    def forward(self, x):
        x = self.conv_layers(x)
        x = x.permute(0,2,1)
        lstm_out,_ = self.lstm(x)
        att = torch.softmax(self.attention(lstm_out), dim=1)
        ctx = torch.sum(att * lstm_out, dim=1)
        return self.classifier(ctx)

# ---------------- load model state safely ----------------
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Device:", device)
model_path = "best_model1_state.pth"
if not os.path.exists(model_path):
    raise FileNotFoundError(f"Put '{model_path}' in the same folder as this notebook and re-run.")
model = ImprovedDeepfakeDetector().to(device)
state = torch.load(model_path, map_location=device)
if isinstance(state, dict) and 'state_dict' in state:
    state = state['state_dict']
new_state = {}
for k,v in state.items():
    new_k = k.replace('module.', '') if k.startswith('module.') else k
    new_state[new_k] = v
model.load_state_dict(new_state)
model.eval()
print("Model loaded ✔\n")

# ---------------- Single file prediction with comprehensive metrics ----------------
print("="*70)
print("SINGLE FILE PREDICTION WITH COMPREHENSIVE ANALYSIS")
print("="*70)
audio_path = r"C:\Users\kevin\Downloads\archive (1)\LA\LA\ASVspoof2019_LA_eval\flac\LA_E_A8746471.flac"
print("Analyzing file:", audio_path)


true_label = None  


try:
    protocol_path = r"C:\Users\kevin\Downloads\archive (1)\LA\LA\ASVspoof2019_LA_cm_protocols\ASVspoof2019.LA.cm.eval.trl.txt"
    if os.path.exists(protocol_path):
        filename = os.path.basename(audio_path)
        file_id = filename.replace('.flac', '')
        with open(protocol_path, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) >= 5 and parts[1] == file_id:
                    label_str = parts[4]  # 'bonafide' or 'spoof'
                    true_label = 1 if label_str == 'bonafide' else 0
                    print(f"Ground truth loaded from protocol: {label_str.upper()}\n")
                    break
except Exception as e:
    print(f"Could not load ground truth: {e}\n")

if not os.path.exists(audio_path):
    raise FileNotFoundError(f"Audio file not found: {audio_path}")

try:
    display(Audio(audio_path, autoplay=False))
except Exception as e:
    print("Audio widget failed:", e)

try:
    # Extract features and predict
    features = extract_features(audio_path)
    print(f"Extracted features shape: {features.shape}")
    tensor = torch.FloatTensor(features).unsqueeze(0).to(device)
    print(f"Tensor shape (batch, channels, frames): {tuple(tensor.shape)}\n")
    
    with torch.no_grad():
        outputs = model(tensor)
        logits = outputs.cpu().numpy()[0]
        probs = torch.softmax(outputs, dim=1).cpu().numpy()[0]
    
    spoof_prob, bonafide_prob = float(probs[0]), float(probs[1])
    pred_label = 1 if bonafide_prob >= 0.5 else 0
    pred_class = "BONAFIDE" if pred_label == 1 else "SPOOF"
    
    # ============ BASIC PREDICTION RESULT ============
    print("="*70)
    print("PREDICTION RESULT")
    print("="*70)
    print(f"Predicted Class: {pred_class}")
    print(f"Bonafide Probability: {bonafide_prob:.4f} ({bonafide_prob*100:.2f}%)")
    print(f"Spoof Probability:    {spoof_prob:.4f} ({spoof_prob*100:.2f}%)")
    print(f"Confidence: {max(spoof_prob, bonafide_prob)*100:.2f}%")
    
    # ============ DETAILED METRICS (if ground truth available) ============
    if true_label is not None:
        true_class = "BONAFIDE" if true_label == 1 else "SPOOF"
        is_correct = (pred_label == true_label)
        
        print(f"\n{'='*70}")
        print("GROUND TRUTH COMPARISON")
        print("="*70)
        print(f"True Label:      {true_class}")
        print(f"Predicted Label: {pred_class}")
        print(f"Result:          {'✓ CORRECT' if is_correct else '✗ INCORRECT'}")
        
        # Confusion matrix elements for single sample
        print(f"\n{'='*70}")
        print("CONFUSION MATRIX INTERPRETATION")
        print("="*70)
        
        if true_label == 0 and pred_label == 0:
            print("Classification: True Negative (TN)")
            print("Description: Correctly identified SPOOF")
        elif true_label == 0 and pred_label == 1:
            print("Classification: False Positive (FP)")
            print("Description: Incorrectly predicted BONAFIDE (was actually SPOOF)")
        elif true_label == 1 and pred_label == 0:
            print("Classification: False Negative (FN)")
            print("Description: Incorrectly predicted SPOOF (was actually BONAFIDE)")
        else:
            print("Classification: True Positive (TP)")
            print("Description: Correctly identified BONAFIDE")
        
        # ============ CLASSIFICATION REPORT ============
        print(f"\n{'='*70}")
        print("CLASSIFICATION REPORT (Single Sample)")
        print("="*70)
        
        # Calculate metrics for single sample
        y_true_single = [true_label]
        y_pred_single = [pred_label]
        
        # Print classification report
        report = classification_report(y_true_single, y_pred_single, 
                                      target_names=['Spoof', 'Bonafide'],
                                      digits=4,
                                      zero_division=0)
        print(report)
        
        # ============ DETAILED METRICS ============
        print(f"{'='*70}")
        print("INDIVIDUAL METRICS")
        print("="*70)
        
        # Accuracy (for single sample: either 1.0 or 0.0)
        accuracy = accuracy_score(y_true_single, y_pred_single)
        print(f"Accuracy:                    {accuracy:.4f}")
        
        # Precision (for Bonafide class)
        if pred_label == 1:
            precision_bonafide = precision_score(y_true_single, y_pred_single, pos_label=1, zero_division=0)
            print(f"Precision (Bonafide):        {precision_bonafide:.4f}")
        else:
            print(f"Precision (Bonafide):        N/A (not predicted as Bonafide)")
        
        # Precision (for Spoof class)
        if pred_label == 0:
            precision_spoof = precision_score(y_true_single, y_pred_single, pos_label=0, zero_division=0)
            print(f"Precision (Spoof):           {precision_spoof:.4f}")
        else:
            print(f"Precision (Spoof):           N/A (not predicted as Spoof)")
        
        # Recall (for Bonafide class)
        if true_label == 1:
            recall_bonafide = recall_score(y_true_single, y_pred_single, pos_label=1, zero_division=0)
            print(f"Recall (Bonafide):           {recall_bonafide:.4f}")
        else:
            print(f"Recall (Bonafide):           N/A (not actually Bonafide)")
        
        # Recall (for Spoof class)
        if true_label == 0:
            recall_spoof = recall_score(y_true_single, y_pred_single, pos_label=0, zero_division=0)
            print(f"Recall (Spoof):              {recall_spoof:.4f}")
        else:
            print(f"Recall (Spoof):              N/A (not actually Spoof)")
        
        # F1-Score (for Bonafide class)
        if pred_label == 1 or true_label == 1:
            f1_bonafide = f1_score(y_true_single, y_pred_single, pos_label=1, zero_division=0)
            print(f"F1-Score (Bonafide):         {f1_bonafide:.4f}")
        else:
            print(f"F1-Score (Bonafide):         N/A")
        
        # F1-Score (for Spoof class)
        if pred_label == 0 or true_label == 0:
            f1_spoof = f1_score(y_true_single, y_pred_single, pos_label=0, zero_division=0)
            print(f"F1-Score (Spoof):            {f1_spoof:.4f}")
        else:
            print(f"F1-Score (Spoof):            N/A")
        
        # Additional useful metrics
        print(f"\nSupport (Actual Class):      {true_class}")
        print(f"Confidence in Prediction:    {max(spoof_prob, bonafide_prob)*100:.2f}%")
    
    # ============ ADDITIONAL ANALYSIS ============
    print(f"\n{'='*70}")
    print("DETAILED ANALYSIS")
    print("="*70)
    print(f"Raw Logits: Spoof={logits[0]:.4f}, Bonafide={logits[1]:.4f}")
    print(f"Logit Difference: {abs(logits[1] - logits[0]):.4f}")
    print(f"Decision Threshold: 0.5000")
    print(f"Distance from Threshold: {abs(bonafide_prob - 0.5):.4f}")
    
    # Confidence interpretation
    confidence = max(spoof_prob, bonafide_prob)
    if confidence >= 0.9:
        conf_level = "Very High"
    elif confidence >= 0.75:
        conf_level = "High"
    elif confidence >= 0.6:
        conf_level = "Moderate"
    else:
        conf_level = "Low"
    print(f"Confidence Level: {conf_level}")
    
    # ============ VISUALIZATIONS ============
    if true_label is not None:
        fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    else:
        fig, axes = plt.subplots(1, 2, figsize=(10, 4))
        axes = [axes[0], axes[1], None]
    
    # Plot 1: Probability Bar Chart
    colors = ['#ff6b6b' if pred_label == 0 else '#95e1d3', 
              '#95e1d3' if pred_label == 1 else '#ff6b6b']
    bars = axes[0].bar(["Spoof", "Bonafide"], [spoof_prob, bonafide_prob], color=colors)
    axes[0].axhline(y=0.5, color='black', linestyle='--', linewidth=1, label='Threshold')
    axes[0].set_ylim(0, 1)
    axes[0].set_ylabel('Probability')
    axes[0].set_title(f'Prediction: {pred_class}', fontweight='bold')
    axes[0].legend()
    
    
    for bar in bars:
        height = bar.get_height()
        axes[0].text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.3f}', ha='center', va='bottom')
    
    # Plot 2: Confidence Gauge
    theta = np.linspace(0, np.pi, 100)
    r = np.ones_like(theta)
    
    ax_polar = plt.subplot(1, 3 if true_label is not None else 2, 2, projection='polar')
    ax_polar.plot(theta, r, 'k-', linewidth=2)
    ax_polar.fill_between(theta, 0, r, alpha=0.3, color='lightgray')
    
    
    colors_gauge = ['red', 'orange', 'yellow', 'lightgreen', 'green']
    n_zones = len(colors_gauge)
    for i, color in enumerate(colors_gauge):
        theta_zone = np.linspace(i * np.pi/n_zones, (i+1) * np.pi/n_zones, 20)
        ax_polar.fill_between(theta_zone, 0, 1, alpha=0.5, color=color)
    
    
    needle_angle = confidence * np.pi
    ax_polar.plot([needle_angle, needle_angle], [0, 0.8], 'r-', linewidth=3)
    ax_polar.plot(needle_angle, 0.8, 'ro', markersize=10)
    
    ax_polar.set_ylim(0, 1)
    ax_polar.set_theta_direction(-1)
    ax_polar.set_theta_zero_location('W')
    ax_polar.set_xticks([0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi])
    ax_polar.set_xticklabels(['0%', '25%', '50%', '75%', '100%'])
    ax_polar.set_yticks([])
    ax_polar.set_title(f'Confidence: {confidence*100:.1f}%\n({conf_level})', 
                      fontweight='bold', pad=20)
    
    # Plot 3: Confusion Matrix (if ground truth available)
    if true_label is not None and axes[2] is not None:
        cm = np.zeros((2, 2))
        cm[true_label, pred_label] = 1
        
        sns.heatmap(cm, annot=True, fmt='.0f', cmap='Blues',
                   xticklabels=['Spoof', 'Bonafide'],
                   yticklabels=['Spoof', 'Bonafide'],
                   ax=axes[2], cbar=False, vmin=0, vmax=1)
        axes[2].set_title('Confusion Matrix (This Sample)', fontweight='bold')
        axes[2].set_ylabel('True Label')
        axes[2].set_xlabel('Predicted Label')
        
        # Highlight the cell
        if is_correct:
            axes[2].add_patch(plt.Rectangle((pred_label, true_label), 1, 1, 
                                           fill=False, edgecolor='green', lw=3))
        else:
            axes[2].add_patch(plt.Rectangle((pred_label, true_label), 1, 1,
                                           fill=False, edgecolor='red', lw=3))
    
    plt.tight_layout()
    plt.show()
    
    # ============ SUMMARY FOR REPORT ============
    print(f"\n{'='*70}")
    print("SUMMARY FOR REPORT")
    print("="*70)
    print(f"File: {os.path.basename(audio_path)}")
    print(f"Prediction: {pred_class}")
    print(f"Confidence: {confidence*100:.2f}%")
    print(f"Bonafide Probability: {bonafide_prob:.4f}")
    print(f"Spoof Probability: {spoof_prob:.4f}")
    if true_label is not None:
        print(f"Ground Truth: {true_class}")
        print(f"Correct: {'Yes' if is_correct else 'No'}")
    
except Exception:
    traceback.print_exc()

# Cleanup
try:
    torch.cuda.empty_cache()
except:
    pass
gc.collect()
print("\n✔ Evaluation complete!")
```

```python
import os, sys, traceback, gc
import numpy as np
import torch
import torch.nn as nn
from sklearn.metrics import (classification_report, confusion_matrix,
                             accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score)
import pandas as pd
import librosa

# ---------------- config (match training) ----------------
class Config:
    SAMPLE_RATE = 16000
    DURATION = 4
    N_MFCC = 60
    N_FFT = 2048
    HOP_LENGTH = 512

config = Config()

# ---------------- safe audio loading + feature extraction ----------------
try:
    import soundfile as sf
except Exception:
    sf = None
try:
    import resampy
except Exception:
    resampy = None


def safe_load_audio(path, sr=config.SAMPLE_RATE, duration=config.DURATION):
    target_len = sr * duration
    y = None; loaded_sr = None

    if sf is not None:
        try:
            data, loaded_sr = sf.read(path, dtype='float32')
            if data is None:
                raise RuntimeError("soundfile returned None")
            if getattr(data, "ndim", 1) > 1:
                data = data.mean(axis=1)
            y = data.astype('float32')
        except Exception as e:
            print("soundfile failed:", e)
            y = None; loaded_sr = None

    if y is None:
        try:
            y, loaded_sr = librosa.load(path, sr=None, mono=True, duration=duration)
            y = y.astype('float32')
        except Exception as e:
            print("librosa load failed:")
            traceback.print_exc()
            raise RuntimeError("Unable to load audio. Install soundfile/libsndfile or ensure ffmpeg is available.") from e

    if loaded_sr != sr:
        if resampy is not None and loaded_sr is not None:
            y = resampy.resample(y, loaded_sr, sr)
            loaded_sr = sr
        else:
            try:
                y = librosa.resample(y, orig_sr=loaded_sr, target_sr=sr)
                loaded_sr = sr
            except Exception:
                raise RuntimeError("Need resampy or librosa resample support to convert sample rate.")

    if len(y) < target_len:
        y = np.pad(y, (0, target_len - len(y)), mode='constant')
    else:
        y = y[:target_len]

    maxv = np.max(np.abs(y)) if y.size else 1.0
    if maxv > 0:
        y = y / maxv
    return y, sr


def extract_features(audio_path, sr=config.SAMPLE_RATE, duration=config.DURATION):
    y, sr = safe_load_audio(audio_path, sr=sr, duration=duration)
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=config.N_MFCC,
                               n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
    mfcc_delta = librosa.feature.delta(mfcc)
    mfcc_delta2 = librosa.feature.delta(mfcc, order=2)
    spectral_centroid = librosa.feature.spectral_centroid(y=y, sr=sr, n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
    spectral_rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr, n_fft=config.N_FFT, hop_length=config.HOP_LENGTH)
    zero_crossing = librosa.feature.zero_crossing_rate(y, hop_length=config.HOP_LENGTH)
    features = np.vstack([mfcc, mfcc_delta, mfcc_delta2, spectral_centroid, spectral_rolloff, zero_crossing])
    return features


class ImprovedDeepfakeDetector(nn.Module):
    def __init__(self, input_dim=config.N_MFCC*3+3, hidden_dim=512, num_layers=4, dropout=0.5):
        super().__init__()
        self.conv_layers = nn.Sequential(
            nn.Conv1d(input_dim,256,3,padding=1), nn.BatchNorm1d(256), nn.ReLU(), nn.MaxPool1d(2), nn.Dropout(dropout*0.5),
            nn.Conv1d(256,512,3,padding=1), nn.BatchNorm1d(512), nn.ReLU(), nn.MaxPool1d(2), nn.Dropout(dropout*0.5),
        )
        self.lstm = nn.LSTM(input_size=512, hidden_size=hidden_dim, num_layers=num_layers,
                            batch_first=True, dropout=dropout if num_layers>1 else 0, bidirectional=True)
        self.attention = nn.Sequential(nn.Linear(hidden_dim*2, hidden_dim), nn.Tanh(), nn.Linear(hidden_dim,1))
        self.classifier = nn.Sequential(
            nn.Linear(hidden_dim*2, hidden_dim), nn.BatchNorm1d(hidden_dim), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(hidden_dim,256), nn.BatchNorm1d(256), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(256,2)
        )
    def forward(self, x):
        x = self.conv_layers(x)
        x = x.permute(0,2,1)
        lstm_out,_ = self.lstm(x)
        att = torch.softmax(self.attention(lstm_out), dim=1)
        ctx = torch.sum(att * lstm_out, dim=1)
        return self.classifier(ctx)


def load_protocol(protocol_path):
    mapping = {}
    try:
        with open(protocol_path, 'r') as f:
            for line in f:
                parts = line.strip().split()
                # protocol format can vary: try to be flexible
                if len(parts) >= 5:
                    file_id = parts[1]
                    label_str = parts[4]
                elif len(parts) >= 2:
                    file_id = os.path.basename(parts[0]).replace('.flac','')
                    label_str = parts[1]
                else:
                    continue
                mapping[file_id] = 1 if label_str.lower().startswith('b') else 0
    except Exception as e:
        print(f"Failed to read protocol {protocol_path}: {e}")
    return mapping



def batch_evaluate(audio_folder, model_path='best_model1_state.pth', protocol_paths=None, out_csv='batch_results.csv', device=None):
    if device is None:
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print('Device:', device)

    if not os.path.exists(model_path):
        raise FileNotFoundError(f"Model file not found: {model_path}")

    
    model = ImprovedDeepfakeDetector().to(device)
    state = torch.load(model_path, map_location=device)
    if isinstance(state, dict) and 'state_dict' in state:
        state = state['state_dict']
    new_state = {}
    for k,v in state.items():
        new_k = k.replace('module.', '') if k.startswith('module.') else k
        new_state[new_k] = v
    model.load_state_dict(new_state)
    model.eval()
    print('Model loaded ✔')

    protocol_map = {}
    if protocol_paths:
        for p in protocol_paths:
            if os.path.exists(p):
                protocol_map.update(load_protocol(p))
    else:
        parent = os.path.dirname(audio_folder)
        candidates = [
            os.path.join(parent, 'ASVspoof2019.LA.cm.eval.trl.txt'),
            os.path.join(parent, 'ASVspoof2019.LA.cm.dev.trl.txt'),
            os.path.join(parent, 'ASVspoof2019.LA.cm.gmm.trl.txt'),
            os.path.join(parent, 'protocol.txt')
        ]
        for c in candidates:
            if os.path.exists(c):
                protocol_map.update(load_protocol(c))

    
    exts = ('.flac', '.wav', '.mp3')
    files = [os.path.join(audio_folder, f) for f in sorted(os.listdir(audio_folder)) if f.lower().endswith(exts)]
    if not files:
        raise FileNotFoundError(f'No audio files found in {audio_folder}')

    rows = []
    y_true = []
    y_pred = []
    y_scores = []

    for i, fp in enumerate(files, 1):
        try:
            filename = os.path.basename(fp)
            file_id = filename.replace('.flac','').replace('.wav','').replace('.mp3','')
            true_label = protocol_map.get(file_id, None)

            features = extract_features(fp)
            tensor = torch.FloatTensor(features).unsqueeze(0).to(device)
            with torch.no_grad():
                outputs = model(tensor)
                probs = torch.softmax(outputs, dim=1).cpu().numpy()[0]
                logits = outputs.cpu().numpy()[0]

            spoof_prob, bonafide_prob = float(probs[0]), float(probs[1])
            pred_label = 1 if bonafide_prob >= 0.5 else 0
            pred_class = 'BONAFIDE' if pred_label==1 else 'SPOOF'
            confidence = max(spoof_prob, bonafide_prob)

            rows.append({
                'filename': filename,
                'file_id': file_id,
                'true_label': true_label if true_label is not None else -1,
                'pred_label': pred_label,
                'spoof_prob': spoof_prob,
                'bonafide_prob': bonafide_prob,
                'confidence': confidence,
                'pred_class': pred_class,
                'logit_spoof': float(logits[0]),
                'logit_bonafide': float(logits[1])
            })

            if true_label is not None:
                y_true.append(true_label)
                y_pred.append(pred_label)
                y_scores.append(bonafide_prob)  # probability for positive class

            if i % 50 == 0 or i==len(files):
                print(f'Processed {i}/{len(files)} files')

        except Exception:
            print(f'Error processing {fp}:')
            traceback.print_exc()

    
    df = pd.DataFrame(rows)
    df.to_csv(out_csv, index=False)
    print('Per-file results saved to', out_csv)

    
    if y_true:
        print('\nAGGREGATED METRICS (on files with protocol labels)')
        print('-------------------------------------------------')
        print('Count:', len(y_true))
        print('Accuracy:', accuracy_score(y_true, y_pred))
        print('Precision (pos=bonafide):', precision_score(y_true, y_pred, pos_label=1, zero_division=0))
        print('Recall (pos=bonafide):', recall_score(y_true, y_pred, pos_label=1, zero_division=0))
        print('F1 (pos=bonafide):', f1_score(y_true, y_pred, pos_label=1, zero_division=0))
        print('\nClassification report:')
        print(classification_report(y_true, y_pred, target_names=['Spoof','Bonafide'], digits=4, zero_division=0))
        cm = confusion_matrix(y_true, y_pred)
        print('Confusion matrix:\n', cm)
        try:
            auc = roc_auc_score(y_true, y_scores)
            print('ROC AUC (bonafide prob):', auc)
        except Exception as e:
            print('ROC AUC could not be computed:', e)
    else:
        print('\nNo ground truth labels were found in protocols. Per-file CSV still created with true_label=-1 for unknowns.')

    low_conf = df.sort_values('confidence').head(10)
    print('\n10 lowest-confidence predictions:')
    print(low_conf[['filename','true_label','pred_label','confidence']].to_string(index=False))

    try:
        torch.cuda.empty_cache()
    except:
        pass
    gc.collect()
    print('\n✔ Batch evaluation complete!')


if __name__ == '__main__':
    audio_folder = r"C:\Users\kevin\Downloads\archive (1)\LA\LA\ASVspoof2019_LA_eval\flac"
    model_path = r"best_model1_state.pth"
    out_csv = r"batch_results.csv"
    protocol_paths = [r"C:\Users\kevin\Downloads\archive (1)\LA\LA\ASVspoof2019_LA_cm_protocols\ASVspoof2019.LA.cm.eval.trl.txt"]

    batch_evaluate(audio_folder=audio_folder, model_path=model_path, protocol_paths=protocol_paths, out_csv=out_csv)
```
